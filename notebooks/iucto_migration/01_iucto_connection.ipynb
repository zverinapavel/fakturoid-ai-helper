# 01 - iÚčto API Connection Test

Test připojení k iÚčto API a načtení základních dat.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

from iucto_migration.iucto_client import IUctoClient
from src.config import config
import os
import json
```

```python
# Initialize iÚčto client
iucto_api_key = os.getenv('IUCTO_API_KEY')

if not iucto_api_key:
    print('❌ IUCTO_API_KEY not found in environment')
    print('Please add to your .env file:')
    print('IUCTO_API_KEY=your_api_key_here')
else:
    iucto = IUctoClient(iucto_api_key)
    print('✓ iÚčto client initialized')
```

```python
# Test connection (upravená buňka 3)
if iucto.test_connection():
    print('✓ Connection to iÚčto successful!')
    
    # Try to get account info (může neexistovat)
    try:
        account = iucto.get_account_info()
        print(f'\nAccount: {account.get("name", "N/A")}')
        print(json.dumps(account, indent=2, ensure_ascii=False))
    except Exception as e:
        print(f'\n⚠️  Account endpoint not available: {e}')
        print('Continuing without account info (this is OK)')
else:
    print('✗ Connection failed')
```

```python
# Load issued invoices from 2013
print('Loading issued invoices from 2013...\n')
issued_2013 = iucto.get_issued_invoices(year=2013, limit=5)

print(f'Found {len(issued_2013)} issued invoices (showing max 5)')
for inv in issued_2013:
    print(f'  - {inv.get("number")}: {inv.get("customer", {}).get("name")} - {inv.get("total")} CZK')
```

```python
# Load received invoices from 2013
print('Loading received invoices from 2013...\n')
received_2013 = iucto.get_received_invoices(year=2013, limit=5)

print(f'Found {len(received_2013)} received invoices (showing max 5)')
for exp in received_2013:
    print(f'  - {exp.get("number")}: {exp.get("supplier", {}).get("name")} - {exp.get("total")} CZK')
```
